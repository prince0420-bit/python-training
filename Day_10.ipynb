```python
def power(num,pov):
  if pov == 0:
    return 1
  else:
    return num * power(num,pov-1)
num=int(input("Enter the number:"))
pov=int(input("Enter the power:"))
result=power(num,pov)
print(result)
```

```python
class bank_account():
    def __init__(self,name,balance,acctype):
        self.name=name
        self.balance=balance
        self.acctype=acctype

    def deposit_amount(self,amount):
       self.balance+=amount
       return self.balance

    def amount_withdrawl(self,amount):
        if self.balance<amount:
            print("insufficient amount")
        else:
            self.balance-=amount
            return self.balance

user1=bank_account("vansh",3000,"saveing account")
withdrawl=user1.amount_withdrawl(2000)
print("amount after withdrawl",withdrawl)
deposit=user1.deposit_amount(3000)
print("amount after deposit",deposit)
```

```python
class Employee():
  def __init__(self, name, department, salary,hike):
    self.name = name
    self.department = department
    self.salary = salary
    self.hike = hike

  def salaryCalculate(self):
    self.salary += self.salary * self.hike / 100


  def showEmployee(self):
    return f"{self.name}, from {self.department}, salary: {self.salary}, hike: {self.hike}"

  def updateDepartment(self, newDepartment):
    self.department = newDepartment

name =  input("Enter the name:")
department = input("Enter the department:")
salary = int(input("Enter the salary:"))
hike = int(input("Enter the hike:"))

employee1 = Employee(name, department, salary, hike)
print(employee1.showEmployee())
employee1.salaryCalculate()
print(employee1.showEmployee())
employee1.updateDepartment("CSE")
print(employee1.showEmployee())
```

```python
class Car():
  def __init__(self, company, model, year_of_manufacture,price,fuel_type,milege):
    self.company = company
    self.model = model
    self.year_of_manufacture = year_of_manufacture
    self.price = price
    self.fuel_type = fuel_type
    self.milege = milege

  def showCar(self):
    return f"{self.company}, {self.model}, {self.year_of_manufacture}, {self.price}, {self.fuel_type}, {self.milege}"

name = input("Enter the company name:")
model = int(input("Enter the model:"))
year_of_manufacture = int(input("Enter the year of manufacture:"))
price = int(input("Enter the price:"))
fuel_type = input("Enter the fuel type:")
milege = int(input("Enter the milege:"))

car1 = Car(name, model, year_of_manufacture, price, fuel_type, milege)
print(car1.showCar())
```
