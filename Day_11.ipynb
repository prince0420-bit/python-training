##Polymorphism
Compile time polymorphism is not present in python
Run time Polymorphism / method overriding

```python
# Polymorphism example
class Abc:
    def calc(self, a, b):
        return a + b

class Def(Abc):
    def calc(self, a, b):
        return a - b

x = Def()
print(x.calc(4, 2))

```

##create a student class and take input name and roll and percentage as the private attribute than set the private attribute
value and print the value of it

```python
class student:
    def __init__(self,name ,roll):
        self.name=name
        self.roll=roll
        self.__percentage=0

    def setPercentage(self,per):
        self.__percentage=per

    def getPercentage(self):
        return f"Percentage {self.__percentage}"

name=input("Enter your Name ")
roll=int(input("Enter your Roll Number "))
s1=student(name,roll)
s1.setPercentage(95)
print(s1.getPercentage())
```

##T-shirt Order Bill
Problem Statement A company is selling T-shirt with three different sizes: Small,Medium and Large.
The prices are fixed:
Rs. 250/Small T-Shirt
Rs. 300/Medium T-Shirt
Rs. 350/Large T-Shirt
Create a program that generates a bill for customers T-Shirt order. The program should ask the user to Enter
the number of small,medium ,large t-shirts they want to purchase.It should then calculate the
total cost and display a formatted bill.
Sample Input 1:
Enter the Number of small T-Shirts: 5
Enter the Number of medium T-Shirts: 3
Enter the Number of Large T-Shirts: 2
Sample Output : Bill Details
Number of small T-Shirts: 5
Number of medium T-Shirts: 3
 Number of Large T-Shirts: 2
Total price = 2850 THANK YOU FOR YOUR ORDER!


```python
class OrderBill:
    def __init__(self,sm,me,lar):
        self.sm=sm
        self.me=me
        self.lar=lar

    def calculateBill_small(self):
        return self.sm*250
    def calculateBill_medium(self):
        return self.me*300
    def calculateBill_large(self):
        return self.lar*350

sm=int(input("Enter the Number of Small T-Shirts: "))
me=int(input("Enter the Number of Medium T-Shirts: "))
lar=int(input("Enter the Number of Large T-Shirts: "))
order=OrderBill(sm,me,lar)

bill1=order.calculateBill_small()
bill2=order.calculateBill_medium()
bill3=order.calculateBill_large()
print(bill1)
print(bill2)
print(bill3)
amount=bill1+bill2+bill3
print("Total price = ",amount," THANK YOU FOR YOUR ORDER! ")
```

##Online Shopping Discount
Problem Statement An online store has a discount scheme. A customer gets a 15% discount on the total Cost
of their order if the total exceeds Rs. 5000. Additionally , if they have a special loyalty coupon, they get an extra
5 % discount on the final discounted price.
Develop a program that calculates the total cost of a customer's order. The program should ask for the unit
price of an item and the quantity they wish to purchase.
It should also ask if they have a loyalty coupon ("y" for yes, 'n for no).

Hint: The price should be printed exactly to two decimal places. If the quantity is less than 1, display "Invalid
quantity". If the coupon input is not "y" or "n", display "Invalid coupon input"
Sample Input 1:
Enter the unit price of the item: 1000
Enter the quantity: 6

```python
unit_price=int(input("Enter the unit price of the item: "))
quantity=int(input("Enter the quantity: "))
coupon=input("Do you have special Loyalty Coupon: ")

if quantity < 1:
    print("Invalid Quantity Entered : ")
else:
    amount = unit_price*quantity
    dis = amount # Initialize dis with the original amount
    if amount > 5000:
        dis = amount - amount*(15/100)
        if coupon.lower() == 'y': # Use lower() to handle both 'Y' and 'y'
            dis = dis - dis*(5/100)
        elif coupon.lower() == 'n': # Use lower() to handle both 'N' and 'n'
            print("No extra Discount is Given: ")
        else:
            print("Invalid Coupon input") # Corrected message based on problem statement
            # We should ideally exit or handle this case, but for now, let's continue with the calculated 'dis' up to this point.
    elif coupon.lower() not in ['y', 'n']: # Handle invalid coupon input when amount is not > 5000
         print("Invalid Coupon input")

    print("Total Amount is : {:.2f}".format(dis)) # Format to two decimal places
```
