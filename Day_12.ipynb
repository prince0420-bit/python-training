# Problem : Calculating the area of circle and a Sector
Problem Statement : Write a program that calculates the area of a circle and ,given a central angle in
 degress , also calculates the area of sector of circle .The program should ask the user to enter the radius
 of the circle and the central angle .The program should have invalid inputs for the radius and angle.
# store the output in a new file "Output".
# SAMPLE INPUT:
# Enter the radius of the circle: 7
# Enter the central angle in degrees: 90
# SAMPLE OUTPUT:
# Area of circle: 153.94
# Area of the Sector: 38.48

```python
file1= open('/content/demo.txt','a')
r = int(input('Enter the radius'))
degree = int(input('enter the angle in degree'))
file1.write(f'Radius: {r}\n')
file1.write(f'Degree: {degree}\n')
AOC = 3.14*(r**2)
file1.write(f'Area of circle: {AOC}\n')
sector = (degree/360)*AOC
file1.write(f'Area of sector: {sector}')
file1.close()
```
