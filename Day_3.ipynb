##calculate the electricity bills based on units consumed


*   first 100 units free
*   after 100 units is 5rs per unit
*   after 200 units is 10rs per unit



```python
units=int(input("enter the units"))
if (units<=100):
    print("electricity is free")
elif (units<=200):
    charge=(units-100)*5
    print(charge)
elif (units>200):
    charge=(100*5)+(units-200)*10
    print(charge)
```

##check if a triangle is valid and determine its type

```python
side1=int(input("enter the side1"))
side2=int(input("enter the side2"))
side3=int(input("enter the side3"))
if (side1==side2==side3):
  print("equilateral triangle")
elif (side1==side2 or side2==side3 or side1==side3):
  print("isosceles triangle")
else:
  print("scalene triangle")
```
