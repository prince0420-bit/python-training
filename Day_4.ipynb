##ask the user for numbers until they enter 0, but only sum positivve numbers using continue

```python
n=1
sum=0
while n!=-1:
    num=int(input("Enter the number or enter 0 to exit "))
    if num==0:
        break
    elif num>0:
        sum=sum+num
        n=n+1
        continue
    else:
        n=n+1

print("Enter Zero  and the total sum is ",sum)
```
