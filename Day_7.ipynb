```python
def swap_characters(s, ch1, ch2):
    result = ""
    for c in s:
        if c == ch1:
            result += ch2
        elif c == ch2:
            result += ch1
        else:
            result += c
    return result

# Example
s = "apples"
ch1 = "a"
ch2 = "p"
print(swap_characters(s, ch1, ch2))
```

```python
def timekeeper(clocks):
    total_hours = sum(clocks)
    highest_hour = max(clocks)
    return total_hours, highest_hour

# Example usage:
clocks = [3, 7, 5, 9]
result = timekeeper(clocks)
print(result[0], result[1])   # Output: 24 9
```

```python
def check_password(password):
    # Condition 1: At least 4 characters
    if len(password) < 4:
        return "Invalid Password: Must be at least 4 characters long."

    # Condition 2: At least 1 digit
    if not any(char.isdigit() for char in password):
        return "Invalid Password: Must contain at least one digit."

    # Condition 3: At least 1 capital letter
    if not any(char.isupper() for char in password):
        return "Invalid Password: Must contain at least one uppercase letter."

    # Condition 4: First character should not be a digit
    if password[0].isdigit():
        return "Invalid Password: First character should not be a digit."

    return "Valid Password"


# Example usage
password = input("Enter your password: ")
print(check_password(password))
```
