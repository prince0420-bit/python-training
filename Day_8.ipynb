##Write the function which take an no. as an input and prints the sum of all it's integer/digit.

```python
def integer(n):
  sum=0
  for digit in str(n):
    sum+=int(digit)
  print(sum)
integer(123)
```

##Write the function for check the no. is prime or not

```python
def check_prime():
    n = int(input("Enter a number: "))
    if n <= 1:
        print(n, "is Not Prime")
        return
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            print(n, "is Not Prime")
            return
    print(n, "is Prime")

check_prime()
```

##Write a function to find the factorial of the no.

```python
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

# Example usage
num = int(input("Enter a number: "))
print("Factorial of", num, "is", factorial(num))  # 120
```
