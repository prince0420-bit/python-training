##Recursion palindrome string or not

```python
def palindrome_String(s):
    if len(s) <=1:
        return True
    if s[0] != s[-1]:
        return False
    return palindrome_String(s[1:-1])

s=input("Enter the String to Check ")
print(palindrome_String(s))
```
