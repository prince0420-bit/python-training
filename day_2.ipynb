```python
text="hello world"

#upper casing
print(text.upper())

#lower casing
print(text.lower())

#capitalize
print(text.capitalize())

#title
print(text.title())

#Swap case
print("HeLLo".swapcase())
```

```python
text="python programming"

#check if start with
print(text.startswith("py"))
print(text.startswith("pro"))

#check if end with
print(text.endswith("ing"))

#find substring(index or -1)
print(text.find("pro"))
print(text.find("java"))

#Find substring
print(text.find("pro"))
print(text.find("java"))

#find substring from right
print(text.rfind("o"))

#index(like find but raise value error if not found)


#Count occurances
print(text.count("o"))

#check if all characters are alphabatic
print(text.isalpha())

#check if all characters are numeric
print(text.isnumeric())

#check if all characters are alphanumeric
print(text.isalnum())

#check if all characters are digit

print("123".isdigit())
print("1.23".isdigit())
```
